```python
"""
Lab 1 - Transparent Rule-Based System
Assumptions:
1. The application is approved only when score >= 70.
2. All required information must be complete.
3. Identity must be verified.
4. Consent must be given.
These are synthetic assumptions for demonstrating transparent rules.
"""

def evaluate_application(score, complete, identity_verified, consent_given):
    """Return a transparent decision and the reasons behind it."""
    reasons = []

    if score < 70:
        reasons.append("Score is below the required threshold of 70.")
    if not complete:
        reasons.append("Required information is incomplete.")
    if not identity_verified:
        reasons.append("Identity verification failed.")
    if not consent_given:
        reasons.append("Consent has not been given.")

    approved = (
        score >= 70
        and complete
        and identity_verified
        and consent_given
    )

    if approved:
        return "APPROVED", ["All four conditions are satisfied."]
    return "NOT APPROVED", reasons


if __name__ == "__main__":
    test_cases = [
        {
            "name": "Case 1 - Pass",
            "score": 85, "complete": True,
            "identity_verified": True, "consent_given": True,
            "expected": "APPROVED"
        },
        {
            "name": "Case 2 - Identity condition fails",
            "score": 85, "complete": True,
            "identity_verified": False, "consent_given": True,
            "expected": "NOT APPROVED"
        },
        {
            "name": "Case 3 - Consent condition fails",
            "score": 85, "complete": True,
            "identity_verified": True, "consent_given": False,
            "expected": "NOT APPROVED"
        },
        {
            "name": "Case 4 - Threshold boundary (score = 70)",
            "score": 70, "complete": True,
            "identity_verified": True, "consent_given": True,
            "expected": "APPROVED"
        },
    ]

    for case in test_cases:
        actual, reasons = evaluate_application(
            case["score"], case["complete"],
            case["identity_verified"], case["consent_given"]
        )
        print(f"\n{case['name']}")
        print(f"Input: score={case['score']}, complete={case['complete']}, "
              f"identity_verified={case['identity_verified']}, "
              f"consent_given={case['consent_given']}")
        print(f"Expected: {case['expected']}")
        print(f"Actual:   {actual}")
        print(f"Reason:   {' '.join(reasons)}")
        assert actual == case["expected"], "Test failed!"
    print("\nAll four test cases passed.")
```
