```python

class ClassroomAgent:
    def __init__(self):
        self.previous_mode = None

    def act(self, temp, occupied):
        # Keep the same thresholds and occupancy priority
        if not occupied:
            target_mode = "ECO"
        elif temp > 26:
            target_mode = "COOL"
        elif temp < 20:
            target_mode = "WARM"
        else:
            target_mode = "IDLE"

        # Send a command only when the target mode changes
        if target_mode == self.previous_mode:
            send_command = False
        else:
            send_command = True
            self.previous_mode = target_mode

        return target_mode, send_command


agent = ClassroomAgent()

# Includes repeated percepts, exact boundaries (20 and 26),
# a mode transition, and an empty-room case.
percepts = [(29, True), (29, True), (20, True), (19, True),
            (19, True), (29, False), (29, False), (24, True)]

print("Step | Previous mode | Percept | Target mode | Command")
for step, percept in enumerate(percepts, 1):
    previous = agent.previous_mode
    target, send = agent.act(*percept)
    command = "SEND" if send else "NO COMMAND"
    print(step, "|", previous, "|", percept, "|", target, "|", command)
```

```python
from lab2_model import ClassroomAgent

agent = ClassroomAgent()

tests = [
    ((29, True), "COOL", True),    # hot and occupied
    ((29, True), "COOL", False),   # repeated target: no new command
    ((20, True), "IDLE", True),    # exact lower boundary
    ((26, True), "IDLE", False),   # exact upper boundary: same mode
    ((19, True), "WARM", True),    # below lower threshold
    ((29, False), "ECO", True),    # empty room has priority
    ((29, False), "ECO", False)    # repeated ECO command avoided
]

for number, test in enumerate(tests, 1):
    percept, expected_mode, expected_send = test
    actual_mode, actual_send = agent.act(*percept)
    passed = actual_mode == expected_mode and actual_send == expected_send
    print("Test", number, ":", "PASS" if passed else "FAIL",
          "| Expected:", expected_mode, expected_send,
          "| Actual:", actual_mode, actual_send)
```
